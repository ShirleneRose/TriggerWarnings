```python
import json
import matplotlib.pyplot as plt

# Load JSON
with open("similar_posts_k5.json") as f:
    data = json.load(f)

# Flatten similarity scores
all_scores = [score for v in data.values() for _, score in v]

# Plot histogram
plt.hist(all_scores, bins=50)
plt.title("Distribution of Cosine Similarity Scores")
plt.xlabel("Similarity")
plt.ylabel("Frequency")
plt.show()
```
