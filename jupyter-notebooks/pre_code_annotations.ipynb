```python
from sentence_transformers import SentenceTransformer
import pandas as pd
import numpy as np

# Load your dataframe
df = pd.read_csv("Combined_Dataset_Annotations - Combined_Dataset.csv")  # Add path to your labeled dataset

# Combine title and body into one text
df["text"] = df["title"].fillna("") + " " + df["body"].fillna("")

# Load model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Compute embeddings
embeddings = model.encode(df["text"].tolist(), show_progress_bar=True)

# Save for later
np.save("post_embeddings.npy", embeddings)
df.to_csv("annotations_dataset_with_index.csv", index=False)  # Save df with aligned index
```
